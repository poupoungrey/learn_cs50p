# Lecture 3: loops

- while keyword/construct to loop
- conditional followed by while then colon and indents to denote the loop
- remember to do base zero counting
- you can use += syntax to add and assign in one operator
- for loops separate from while and can run across a list instead of incrementing
- list data structure
- if you need a pure counter for a for loop, use _ instead of 1. only if you don't care about actually recording the value. signals to readers that this is purely a counter
- py will not stop you from creating infinite loops
- can use while and a condition with a break outcome 
- indexing into a list using square brackers, ie foobar_list[0]
- dict data structure = associating one value with another, uses curl brackets
- can have a list where each element is a dictionary
- 'None' keyword denotes absense of a value


```python
i = 0
while i <= 2: 
    print("meow")
    i += 1

for i in [0, 1, 2]:
    print("moo")

test_range = range(5)
print(test_range)

for _ in range(3):
    print("woof") 

print("bro\n" * 3, end = "")

aa =list[1, 2, 3, 5, 7]
print(aa)

bros = ["trav", "coop", "slob"]
for bro_number in bros: 
    print(bro_number)

bros = {
    "trav": "fratty", 
    "coop": "fratty", 
    "slob": "pledge"
}
for bro in bros:
    print(bro, bros[bro])
```

# Lecture 4: exceptions

- runtime errors happen when program is being used and have to code defensively to prevent errors
- keyword 'try' to attempt an action, then after the indendented code use keyword 'except' to perform alternative action
- should ideally only use one line of code ish in a try statement
- error types: NameError, ValueError, IndexError, SyntaxError
- can use 'else' keyword on a try/except statement to perform execution if no errors occur
- 'return' also breaks out of a loop
- py less focused on if statement checks and it is preferred to use 'try' and hope it works and then handle error cases



# Lecture 5: libraries

# Lecture 6: Unit Tests

- can create a testing file, which imports the func to be tested via: from filename import functotest

# Lecture 7: File I/O

- 'append' function will add an element at the end of a list, like list_name.append(3)
- 'open' function to open a file in py, just need to specify the file name and how to open
- one argument is "w" which means open the file with the intention to write to it (this will create the file if it does not already exist)
- but "w" rewrites/overwrites the file each time, so use "a" to append
- file_variable.write(string_variable) to write something to a file, then file_variable.close() to complete the interaction with the file and remove the handle (maybe this also removes a latch/lock as well)
- use 'with' statements when writing to files so you dont have to use close
- file_object.readLines() to read the file contents and bring them into memory as a list
- functions can be passed as arguments of functions in python
- lambda functions aka anonymous functions can be used, literally with keyword 'lambda'
- use the csv module/package for optimal reading of these files
- csv functions include csv.reader(file) and the reader will handle escape characters and seperators etcetera
- there are also dictionary readers and there are methods for writing
- pillow library for reading image files and gifs (import PIL)
