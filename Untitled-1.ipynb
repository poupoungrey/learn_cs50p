# Lecture 8: Regular expressions

- regex library is re
- re.search() to search for a pattern in a string
- finite state machines = reflexive edges decide to stay in an evaluated state and either remain or transition to another state and then evaluate if it changed state or not
- backslash escape character to do literals
- use prefix r before reg expression to clarify/confirm it as a pattern and not just a string (ie, r'regex[a-z]')
- use a library to validate emails, don't try to write your own regex
- case sensitive by default, you have to ignore case via a flag argument
- can use groups in match to ultimately extract/parse the parts of the string for use in later
- new walrus operator ( := ) added to python that will both assign an object to a variable name and also run a boolean comparison operator


```python
import re

emails = ['hi@bro.com', 'usug99@bro.com', 'invalid@', 'broken.com', '@', 
          '@bro.com', 'fake@brocom', 'hi@bro.combro', 'hi@@bro.com', 'hi@bro.edu', 
          '.com@bro.com', 'a&a@bro.com', '_@bro.com', 'hi@&.com', 'hi@bro.COM']

for i in emails:
    print(re.search(pattern='@', string=i))

print('\n')

for i in emails:
    print(re.search(pattern=r'^[^@]+@[^@]+\.com$', string=i))

print('\n')

for i in emails:
    print(re.search(pattern=r'^[a-zA-Z0-9_]+@[a-zA-Z0-9]+\.com$', string=i, flags=re.IGNORECASE))

```

# Lecture 9: Object Oriented Programming

- tuple data structure is like a list but difference is it is immutable. (x, y, z, ...)
- if you try to change a tuple value you will get a TypeError
- a class is a programmer defined data type (a key component of any OOP lang)
- 'class' keyword to denote a class
- by convention classes are named with capital first letter
- you create objects from classes
- classes are like blueprints, objects are constructions made from the blueprint
- can define methods for classes, sort of like behaviors or acceptable functions that can be run on the class
- def __init__(self, ): is what you have to type inside the class in order for the constructor to run the class and create an object of the class
- classes support better type checking, validation, error handling
- 'raise' keyword to create your own exceptions/errors
- __str__ method to how to print your object as a string
- when you create a method for a class, always have at least one argument called 'self'
- properties are attributes that have more guardrails/controls: @property keyword
- getters and setters (gets a value) (sets a value)
- a getter is designated as @property and a setter is defined with @property_name.setter
- python has a lot of conventions but not a lot of constraints, making it potentially vulnerable to bad coding practices or bugs
- @classmethod does not have access to self but can be used for all objects every constructed from the class. these are different from instance methods
- in a classmethod, first argument is 'cls' which is a reference to the class itself
- inheritance: feature where classes can share elements instead of reimplementing the same thing in multiple classes
- to denote inheritence, in the definition of the inheriting class put the name of the super class in the parenthesis. IE, def Class1(Superclass): 
- then, under __init__, super().__init__(arg) to denote where something should be inherited from
- operator overloading to create custom operations for a class


```python
import random

print(type(5))
print(type("goodbye planet"))
print(type([1, 3]))

class Frat: 

    chapters = ['dz', 'tnt', 'yyy', 'gdi']

    @classmethod
    def pick(cls, pledge): 
        print(pledge, 'is in', random.choice(cls.chapters))


Frat.pick('Xander')
Frat.pick('Kyle')
Frat.pick('Travis')
```
