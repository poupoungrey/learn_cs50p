# Lecture 3: loops

- while keyword/construct to loop
- conditional followed by while then colon and indents to denote the loop
- remember to do base zero counting
- you can use += syntax to add and assign in one operator
- for loops separate from while and can run across a list instead of incrementing
- list data structure
- if you need a pure counter for a for loop, use _ instead of 1. only if you don't care about actually recording the value. signals to readers that this is purely a counter
- py will not stop you from creating infinite loops
- can use while and a condition with a break outcome 
- indexing into a list using square brackers, ie foobar_list[0]
- dict data structure = associating one value with another, uses curl brackets
- can have a list where each element is a dictionary
- 'None' keyword denotes absense of a value


```python
i = 0
while i <= 2: 
    print("meow")
    i += 1

for i in [0, 1, 2]:
    print("moo")

test_range = range(5)
print(test_range)

for _ in range(3):
    print("woof") 

print("bro\n" * 3, end = "")

aa =list[1, 2, 3, 5, 7]
print(aa)

bros = ["trav", "coop", "slob"]
for bro_number in bros: 
    print(bro_number)

bros = {
    "trav": "fratty", 
    "coop": "fratty", 
    "slob": "pledge"
}
for bro in bros:
    print(bro, bros[bro])
```
